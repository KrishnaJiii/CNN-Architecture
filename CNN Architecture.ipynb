Question 1: What is the role of filters and feature maps in Convolutional Neural Network (CNN)?

Answer 1: In a Convolutional Neural Network (CNN), filters and feature maps play a key role in extracting important information from input images. A filter (also called a kernel) is a small matrix that slides over the input image and performs convolution to detect specific features such as edges, corners, or textures. Each filter is designed to focus on a particular pattern in the image. When a filter moves across the image, it computes values based on the matching patterns, helping the network understand local features.

The result of applying a filter to the input image is called a feature map. A feature map shows where a particular feature is present in the image and how strongly it appears. For example, if a filter is trained to detect vertical edges, its feature map will highlight the regions where vertical edges exist. By using multiple filters, a CNN generates multiple feature maps, allowing it to learn different features at various levels of complexity. Together, filters and feature maps help the CNN automatically learn and represent important visual patterns needed for tasks like image classification and object detection.

Question 2: Explain the concepts of padding and stride in CNNs(Convolutional Neural Network). How do they affect the output dimensions of feature maps?

Answer 2: In Convolutional Neural Networks (CNNs), padding and stride are important concepts that control how a filter moves over the input image and how the output feature map is formed. Padding means adding extra pixels, usually zeros, around the border of the input image before applying the filter. The main purpose of padding is to preserve the spatial size of the image and prevent loss of important information at the edges. For example, with proper padding, the output feature map can have the same height and width as the input image, which is often called same padding.

Stride refers to the number of pixels by which the filter moves each time it slides over the input image. A stride of 1 means the filter moves one pixel at a time, producing a larger feature map with more detailed information. A larger stride, such as 2, makes the filter jump two pixels at a time, resulting in a smaller feature map. In simple terms, padding increases or maintains the output size, while stride reduces the output size. Together, padding and stride directly affect the dimensions of the feature maps and help control the level of detail and computation in a CNN.

Question 3: Define receptive field in the context of CNNs. Why is it important for deep architectures?

Answer 3: In a Convolutional Neural Network (CNN), the receptive field is the part of the input image that affects the output of a particular neuron. In early layers, the receptive field is small and focuses on local features like edges or corners. As the network becomes deeper, the receptive field increases, allowing neurons to capture larger and more meaningful patterns. This is important because deep CNNs can then understand both simple and complex features, helping in accurate image classification and recognition.

Question 4: Discuss how filter size and stride influence the number of parameters in a CNN.

Answer 4: In a Convolutional Neural Network (CNN), the filter size and stride affect the number of parameters and the computation of the network. A larger filter size contains more weights, which increases the number of parameters and makes the model more complex. For example, a 5×5 filter has more parameters than a 3×3 filter. On the other hand, stride does not directly change the number of parameters in a filter, but a larger stride reduces the size of the output feature map, which lowers the overall computation and number of activations. Thus, filter size controls model complexity, while stride controls the spatial size of feature maps.

Question 5: Compare and contrast different CNN-based architectures like LeNet, AlexNet, and VGG in terms of depth, filter sizes, and performance.

Answer 5: LeNet, AlexNet, and VGG are popular CNN architectures that differ in depth, filter sizes, and performance. LeNet is a shallow network with only a few layers and mainly uses larger filters like 5×5, making it suitable for simple tasks such as handwritten digit recognition. AlexNet is deeper than LeNet and uses a combination of large and small filters, with 11×11 filters in early layers, and it performs much better on complex image datasets. VGG is a very deep network that uses only small 3×3 filters stacked together, which improves feature learning and accuracy. Overall, as depth increases from LeNet to VGG, performance improves, but computational cost also increases.

```python
# Import required libraries
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.utils import to_categorical

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Reshape and normalize data
x_train = x_train.reshape(-1, 28, 28, 1) / 255.0
x_test = x_test.reshape(-1, 28, 28, 1) / 255.0

# One-hot encode labels
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)

# Create CNN model
model = Sequential()
model.add(Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)))
model.add(MaxPooling2D((2,2)))
model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dense(10, activation='softmax'))

# Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
model.fit(x_train, y_train, epochs=5, batch_size=32)

# Evaluate the model
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print("Test Accuracy:", test_accuracy)
```

Question 8: Using PyTorch, write a script to define and train a CNN on the MNIST dataset. Include model definition, data loaders, training loop, and accuracy evaluation

```python
# Import required libraries
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Device configuration
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Data preprocessing and loading
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

train_dataset = datasets.MNIST(root='./data', train=True,
                               transform=transform, download=True)
test_dataset = datasets.MNIST(root='./data', train=False,
                              transform=transform, download=True)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)

# CNN model definition
class CNN(nn.Module):
    def __init__(self):
        super(CNN, self).__init__()
        self.conv1 = nn.Conv2d(1, 32, 3)
        self.pool = nn.MaxPool2d(2, 2)
        self.conv2 = nn.Conv2d(32, 64, 3)
        self.fc1 = nn.Linear(64 * 5 * 5, 128)
        self.fc2 = nn.Linear(128, 10)

    def forward(self, x):
        x = self.pool(torch.relu(self.conv1(x)))
        x = self.pool(torch.relu(self.conv2(x)))
        x = x.view(-1, 64 * 5 * 5)
        x = torch.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# Initialize model, loss, and optimizer
model = CNN().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training loop
for epoch in range(5):
    model.train()
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")

# Accuracy evaluation
model.eval()
correct = 0
total = 0
with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total
print("Test Accuracy:", accuracy)
```
